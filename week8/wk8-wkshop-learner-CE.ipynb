```python
# Part 2: Data Cleaning

# 1. Handling Missing Values
```

```python
# Create copies of original DataFrames for cleaning
customers_copy = customers.copy()
products_copy = products.copy()
stores_copy = stores.copy()
sales_copy = sales.copy()
inventory_copy = inventory.copy()
```

```python
# For numerical columns with missing values, replace with column mean
products_copy['weight'] = products_copy['weight'].fillna(products_copy['weight'].mean())
stores_copy['size_sqft'] = stores_copy['size_sqft'].fillna(stores_copy['size_sqft'].mean())
inventory_copy['reorder_level'] = inventory_copy['reorder_level'].fillna(inventory_copy['reorder_level'].mean())
```

```python
# For categorical columns with missing values, replace with most frequent value
sales_copy['payment_method'] = sales_copy['payment_method'].str.strip().str.upper()
sales_copy['payment_method'] = sales_copy['payment_method'].replace({
    'CC': 'CREDIT CARD',
    'CREDIT': 'CREDIT CARD',
    'DC': 'DEBIT CARD',
    'DEBIT': 'DEBIT CARD',
    'PP': 'PAYPAL',
    'APPLEPAY': 'APPLE PAY',
    'GOOGLEPAY': 'GOOGLE PAY',
    'VISA': 'CREDIT CARD'
})
sales_copy['payment_method'] = sales_copy['payment_method'].fillna(sales_copy['payment_method'].mode()[0])
```

```python
# Categorical placeholder used used for free-text identifiers, as mode not applicable
customers_copy['email'] = customers_copy['email'].fillna('Unknown')
customers_copy['phone'] = customers_copy['phone'].fillna('Unknown')
```

```python
# zip_code: assessed from address, city, and state already provided in customers table
customers_copy['zip_code'] = customers_copy['zip_code'].fillna('78757')
# "135 Madison St", city Austin, state TX. corresponds to 78757 zip code
```

```python
# For date columns with missing values, use forward fill or backward fill as appropriate
# These load as text, not dates - convert first
customers_copy['registration_date'] = pd.to_datetime(customers_copy['registration_date'])
stores_copy['opening_date'] = pd.to_datetime(stores_copy['opening_date'])
inventory_copy['last_restock_date'] = pd.to_datetime(inventory_copy['last_restock_date'])
sales_copy['date'] = pd.to_datetime(sales_copy['date'])

# Only sales.date gets filled. The other three aren't in date order, so a fill
# would just copy an unrelated record's date. Leaving those null.
```

```python

# Sales copy dates are in order, so forward fill is appropriate
sales_copy['date'] = sales_copy['date'].ffill()
```

```python
# sales.store_id -> 1 missing (sale_id 11)
# Date, customer_id, and product_id are present identifiers but they do not allow us to determine which store the sale occurred at
# Therefore, best to leave null
sales_copy[sales_copy['store_id'].isnull()]
```

```python
# State written two ways (TX vs Texas)
STATE_MAP = {'California':'CA','Texas':'TX','North Carolina':'NC','Washington':'WA',
             'Oregon':'OR','Maryland':'MD','Massachusetts':'MA','Georgia':'GA'}
customers_copy['state'] = customers_copy['state'].str.strip().replace(STATE_MAP)
stores_copy['state'] = stores_copy['state'].str.strip().replace(STATE_MAP)
customers_copy['state'].value_counts()
```

```python
# Missing value final checks
print(customers_copy.isnull().sum())
print(stores_copy.isnull().sum())
print(inventory_copy.isnull().sum())
print(sales_copy.isnull().sum())
print(products_copy.isnull().sum())

# Remaining nulls are the date and ID fields with no valid fill - see above
```

```python
# 2. Removing Duplicates
# Check for exact duplicate rows
print("Customers exact duplicates:", customers_copy.duplicated().sum())
print("Products exact duplicates:", products_copy.duplicated().sum())
```

```python
# Check for duplicates based on identifying columns (not just exact row matches)
print("Customers duplicate customer_id:", customers_copy.duplicated(subset=['customer_id']).sum())
print("Products duplicate product_id:", products_copy.duplicated(subset=['product_id']).sum())
```

```python
# View any duplicated rows if found
customers_copy[customers_copy.duplicated(subset=['customer_id'], keep=False)]
```

```python
# Remove duplicate rows
customers_copy = customers_copy.drop_duplicates(subset=['customer_id'], keep='first')
customers_copy[customers_copy.duplicated(subset=['customer_id'], keep=False)]
```

```python
# Approach for identifying duplicates:
# Started with exact row matches - none in either table
# Then checked customer_id and product_id, since the same person or product can be entered twice with different formatting. Found one: customer 26
# Also checked products on name+brand (nothing) and customers on phone digits
# Phone flagged 3, but two of those are different people sharing a number, not duplicates. 
#   Worth noting duplicated() counts NaN as a match, so nulls have to be dropped first or you get false positives.
# Kept the first Betty Lewis row. Address/phone/state are just formatting differences, but the two emails are actually different 
#   - so this does throw away a real email address.
```

```python
#   Discussion Question 1
#   Key advantages of using Pandas for data cleaning compared to other methods
#
#   Vectorized .fillna()/.isnull()/.duplicated() apply a rule across an entire column or row-set in one line - no manual looping through rows
#   .isnull().sum() and .duplicated().sum() give instant column- and row-level audits, surfacing exactly which columns and how many rows are affected in seconds rather than viual scan
#   Chaining (.str.upper(), .map(), .value_counts()) lets you normalize inconsistent categorical entries before filling, and separate true duplicates from formatting variants 
#       both require inspecting content, not just presence, which Pandas makes cheap to do at scale
```

```python
#   Discussion Question 2
#   How would your approach to handling missing values differ if the missing data was not random but had a pattern or meaning?
#
#   Not every null - or duplicate - should be treated the same way:
#   Check row order before trusting a forward/backward fill; if the order isn't confirmed chronological, a fill can silently fabricate a value that never existed
#   Distinguish externally recoverable information from internally unrecoverable information: some missing fields correspond to real-world facts that exist independently of the dataset and can be looked up
#       (an address implies a real zip code); others have no external truth to reference at all - only whatever correlating data exists inside the dataset itself, and if that comes up empty, there's nothing to recover
#   Distinguish a genuine data conflict from a cosmetic one when resolving duplicates: differences that are just formatting variants of the same underlying value are a non-issue, but a difference in actual content
#       (e.g., two different real values for the same field) forces an explicit choice about what to keep and what's lost
```

```python
#   Discussion Question 3: What types of data quality issues might not be
#   immediately visible through simple DataFrame inspection methods?
#
#   Presence/uniqueness checks don't catch correctness:
#   Categorical variants (typos, case, abbreviations) look like distinct values, not missing ones - need normalizing before mode-fill
#   .duplicated() treats NaN == NaN as a match, so nulls cause false-positive duplicate counts - filter nulls before checking
#   A value can be present, unique, and still wrong (negative price, cost > price) - isnull()/duplicated() miss this; needs a reasonableness check instead
```
